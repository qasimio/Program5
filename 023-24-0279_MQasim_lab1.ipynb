```python
# Task 2

def is_prime():
    num = int(input("Enter a number(to check if prime): "))
    for x in range(2, num):
        remain = num % x
        if remain == 0:
            print(f"{num} is not a prime number")
            break
    else:
        print(f"{num} is a prime number")

is_prime()
```

```python
# Task 3 - Febnocchi sequence

def fibonacci(n):
    x = 0
    y = 1
    for _ in range(n):
        z = x+y
        x =y
        y = z
        print(z, end=" ")

fibonacci(10)
```

```python
# Task 4 - ordered unique list


def uniquelist(olist):
    nlist = []
    for item in olist:
        if item not in nlist:
            nlist.append(item)

    return nlist


olist = list(map(int, input("Enter numbers separated by space: ").split()))

uniquelist(olist)

```

```python
# Task 5

def multiply(*args):

    num = 1

    for item in args:
        num *= item

    return num

multiply(1, 3, 5, 8, 9)
```

```python
#  Task 6

text = input("Enter a string: ")
frequency = {}

for char in text: 
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
print(frequency)
```

# matplotlib

```python
# Task 1
import pandas as pd
import matplotlib.pyplot as plt

students = pd.DataFrame({
    "name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman", "Hamza", "Fatima", "Zain"],
    "section": ["A", "A", "B", "B", "A", "B", "A", "B"],
    "marks": [85, 72, 45, 91, 66, 38, 79, 55]
})

section_marks = students.groupby("section")["marks"].mean()

plt.bar(section_marks.index, section_marks.values)
plt.xlabel("Section")
plt.ylabel("Average Marks")
plt.title("Average Marks per Section")
plt.show()
```

```python
# Task 2
plt.hist(students["marks"], bins=5)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

```python
# Task 3
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(students["name"], students["marks"], marker="o")
axes[0].set_title("Student Marks")
axes[0].set_xlabel("Student")
axes[0].set_ylabel("Marks")

axes[1].scatter(students["marks"], range(len(students)))
axes[1].set_title("Marks Scatter Plot")
axes[1].set_xlabel("Marks")
axes[1].set_ylabel("Student Number")

plt.tight_layout()
plt.show()
```
